**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1

class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


s1 = Student("Ahmed", 101)

print("Name:", s1.get_name())
print("Roll Number:", s1.get_roll_number())

s1.set_name("Ali")
s1.set_roll_number(102)

print("Updated Name:", s1.get_name())
print("Updated Roll Number:", s1.get_roll_number())

```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawal successful")

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

account.deposit(2000)
account.withdraw(1500)

print("Remaining Balance:", account.get_balance())
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Write your code here for Q3

class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary


emp = Employee(30000)

print("Salary:", emp.get_salary())

emp.set_salary(-5000)
print("Salary:", emp.get_salary())

emp.set_salary(40000)
print("Updated Salary:", emp.get_salary())
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Write your code here for Q4

class Example:
    def __init__(self):
        self._protected = "Protected Variable"
        self.__private = "Private Variable"

    def show_private(self):
        print(self.__private)


obj = Example()

# Protected variable can be accessed by convention
print(obj._protected)

# Private variable is accessed through a method
obj.show_private()
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


n1 = Number(10)
n2 = Number(20)

result = n1 + n2

print("Sum:", result.value)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Write your code here for Q16

from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


# Try to create an object
try:
    c = Computer()
except TypeError as e:
    print("Error:", e)


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data")


l = Laptop()
l.process()
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing data")


laptop = Laptop()
desktop = Desktop()

laptop.process()
desktop.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18

from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass


class ElectricCar(Car):
    def drive(self):
        print("Electric car is driving")


class SportsCar(Car):
    def drive(self):
        print("Sports car is driving")


car1 = ElectricCar()
car2 = SportsCar()

car1.start_engine()
car1.drive()

car2.start_engine()
car2.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20

from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")


class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI")


def process_payment(payment_method, amount):
    payment_method.pay(amount)


credit_card = CreditCardPayment()
upi = UPIPayment()

process_payment(credit_card, 5000)
process_payment(upi, 2000)
```
